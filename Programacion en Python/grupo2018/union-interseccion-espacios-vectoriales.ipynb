Sin embargo con el sistema intersección es algo más complejo. Para encontrar la base tenemos primero que expresar los subespacios con sus ecuaciones implícitas. Vamos a ilustrarlo con un ejemplo. Sean $a_1$ y $a_2$ dos vectores de $\mathbb{R}^4$, donde $a_1 = (1,0,3,-2)$ y $a_2 = (2,2,1,-2)$.

$$
  a_1 = \left(\begin{array}{r}
1\\
0\\
3\\
-2\\
\end{array}\right)\,\,
  a_2 = \left(\begin{array}{r}
2\\
2\\
1\\
-2\\
\end{array}\right)
$$

Simbólicamente:

$$
  \left[ \begin{array}{cc|c}
  1 & 2 & x_1 \\
  0 & 2 & x_2 \\
  3 & 1 & x_3 \\
  -2 & -2 & x_4 \\
  \end{array}  \right]
  \rightarrow
  \left[ \begin{array}{cc|c}
  1 & 2 & x_1 \\
  0 & 2 & x_2 \\
  0 & 0 & -3 x_1 - \frac{5}{2} x_2 + x_3 \\
  0 & 0 & 2 x_1 - x_2 + x_4 \\
  \end{array}  \right]
$$

Con lo que las ecuaciones implícitas quedarían como:

$$
  0 = -3 x_1 - \frac{5}{2} x_2 + x_3 \\
  0 = 2 x_1 - x_2 + x_4 \\
$$

No hace falta complicar más de lo necesario, así que no vamos a trabajar con operaciones simbólicas. Usaremos una representación matricial también para las ecuaciones implícitas.

$$
  \left[ \begin{array}{rr|cccc}
  1 &  2 & 1 & 0 & 0 & 0 \\
  0 &  2 & 0 & 1 & 0 & 0 \\
  3 &  1 & 0 & 0 & 1 & 0 \\
 -2 & -2 & 0 & 0 & 0 & 1 \\
  \end{array}  \right]
  \rightarrow
  \left[ \begin{array}{cc|cccc}
  1 & 2 & 1 & 0 & 0 & 0\\
  0 & 2 & 0 & 1 & 0 & 0 \\
  0 & 0 & -3 & -\frac{5}{2} & 1 & 0 \\
  0 & 0 & 2 & -1 & 0 & 1 \\
  \end{array}  \right]
$$

$$
