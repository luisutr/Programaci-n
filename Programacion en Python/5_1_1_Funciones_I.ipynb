## 1. Descomposición y abstracción

Las dos características principales de las funciones son:

* **Descomposición**:
    * Permite dividir un problema en problemas más pequeños y fáciles de resolver
    * Es posible reutilizar funciones que resuelven un problema dado
    * Las funciones aportan estructura al código
* **Abstracción**:
    * Cada función permite ocultar el detalle, como si fuese una caja negra
    * La información relevante será visible
    * La información irrelevante se invisibiliza

Las funciones permiten crear objetos computacionales complejos que se comportan como primitivas. Esto lo consiguen proporcionando un mecanismo para descomponer y abstraer.

La descomposición crea estructura. Permite romper el problema en subproblemas que son razonablemente auto-contenidos y que podrían reutilizarse en otros contextos.

La abstracción oculta detalle haciendo ver el código como una **caja negra**. Es decir, algo cuyo interior no puede verse, ni se tiene que ver, ni debe verse.  La esencia de la programación es preservar la información relevante en cada contexto mientras se oculta la no relevante. La clave es encontrar la relación de relevancia que es apropiada tanto para los clientes como para el creador de la abstracción. En esto consiste el arte de programar de manera efectiva.

A continuación se presenta un programa principal con unas funciones que resuelven subproblemas:

```python
#PROGRAMA QUE SOLICITA POR TECLADO VARIOS NÚMEROS SEPARADOS POR COMAS Y DEVUELVE
#EL PROMEDIO DE LOS NÚMEROS Y UNA LISTA CON LOS MAYORES AL PROMEDIO
#SI NO SE INDICA NINGÚN VALOR A LA ENTRADA SE OBTIENE UN MENSAJE INDICÁNDOLO

#Función para convertir un string a lista de números
def convertir_a_lista_numeros(cadena):
  if cadena == "":
    return []
  else:
    lista_numeros = cadena.split(",")
    lista_numeros_sin_espacios = []
    for i in lista_numeros:
       lista_numeros_sin_espacios.append(float(i))
    return lista_numeros_sin_espacios
    #return [float(i) for i in cadena.split(",")]

# Función para calcular el promedio de una lista de números
def calcular_promedio(numeros):
    if len(numeros) == 0:
        return None  # Retornamos None si la lista está vacía
    return sum(numeros) / len(numeros)

# Función para encontrar los números mayores que el promedio
def mayores_que_promedio(numeros):
    promedio = calcular_promedio(numeros)
    if promedio is None:
        return None, []  # Retornamos None y una lista vacía si no hay datos
    mayores = [num for num in numeros if num > promedio]
    return promedio, mayores

# Lógica principal del programa
print("Análisis de números")
cadena_numeros = input("Introduce una lista de números separados por comas: ")
numeros = convertir_a_lista_numeros(cadena_numeros)
promedio, mayores = mayores_que_promedio(numeros)

if promedio is None:
    print("La lista de números está vacía. No se puede calcular el promedio.")
else:
    print(f"El promedio de los números es: {promedio:.2f}")
    print(f"Los números mayores que el promedio son: {mayores}")
```

## 3. Expresión de llamada

La expresión de llamada se utiliza para ejecutar una función previamente definida:
* Llamar, invocar, usar una función.
* Incluye el formato básico: **nombre_funcion(parametros_reales)**
* La función debe estar definida para poder llamarla/ejecutarla
* Los parámetros reales son los valores para los que se ejecutará la función
* **Asociación de parámetros reales a parámetros formales**
* Creación de los parármetros formales como variables
* Tras terminar la ejecución, el **punto de ejecución** se transfiere a la sentencia que haya tras la llamada a la función, eliminando las variables creadas en la función (incluyendo los parámetros formales)
* **El valor retornado en la función, se asocia a la expresión de llamada**

La **asociación de parámetros reales a los formales** se realiza cuando se llama a una función (se invoca, se llama): **primero se evalúan las expresiones para construir los valores de los parámetros o argumentos**. Entonces, a los **parámetros formales se le asocian a los parámetros reales** que se le han pasado (que se han escrito entre paréntesis en la expresión de llamada). Este proceso de asociación es similar a la asignación.

Entonces, el **punto de ejecución** (la instrucción siguiente a ejecutar) cambia a la primera instrucción del cuerpo de la función, y el código de la función se va ejecutando hasta que se encuentra una sentencia `return` o terminan las sentencias del cuerpo. Si se acaban las sentencias la función devuelve `None`. También ocurre lo mismo cuando no hay expresión después de `return`.

El valor retornado es el valor de la expresión de llamada.

Al terminar, el **punto de ejecución** se transfiere de nuevo al código que viene inmediatamente a continuación de la llamada.

## 5. Ámbito (*scope*) y ciclo de vida de las variables
* El **ámbito de una función define el contexto en el que dicha función es visible y tiene significado válido**.
* Una función crea su propio **espacio de nombres** al ser invocada y éste se destruye al concluir su ejecución.
* Las variables definidas en una función son válidas solo cuando se está ejecutando dicha función y serán destruidas cuando termine su ejecución.
* **Al ejecutar una función también son válidas las variables del ámbito superior a la función, siempre que no se redefinan dentro de ésta.**

Al salir del intérprete o sesión de Python, las definiciones realizadas (funciones y variables) se pierden.

### Tipos de variables:
* **Locales**: creadas en un bloque con alcance únicamente en él
* **Globales**: no definida en ningún bloque particular. Es posible visualizarla y modificarla (indicando que es una variable global) en cualquier bloque. Es preferible no usarlas en funciones, sino pasar los valores necesarios en los argumentos

```python
#Ejemplo de variables locales y globales
def f(x):
    y = 1
    x = x + y
    print('x =', x)
    return x
x = 3; y = 2
z = f(x)
print('z =', z)
print('x =', x)
print('y =', y)
```

En el nivel superior, fuera de cualquier función, el intérprete de Python mantiene una tabla de símbolos con los nombres asignados a los objetos correspondientes.

Cuando se llama a una función se crea una nueva tabla de símbolos (se llama normalmente **marco de pila**). Esta tabla mantiene los símbolos definidos dentro de la función, incluidos los parámetros formales.  Si desde esta función se llama a otra, se crea una nueva tabla de símbolos y así sucesivamente.

Cuando una función termina, la tabla de símbolos correspondiente deja de existir. En Python se puede saber el ámbito de una variable simplemente mirando al texto del programa. Ésto se conoce como ámbitos léxicos o estáticos.


```python
def f():
    print(x)   #se refiere a la variable global

def g():
    print(x)  #se refiere a la variable local por la asignación siguiente
    x = 1

x = 3
f()
x = 4
g()
```

La asignación de `x` en la función `g` hace que el símbolo `x` se considere un nombre local de `g`.  Por tanto, su uso antes de la asignación se considera un error.  El mismo código que en `f` no genera ningún problema, porque se refiere a la variable global `x`. En `g` es un error porque se refiere a la variable local `x` que aún no ha sido asignada.

Si se quiere usar el valor de una variable global y modificarla dentro de una función, se debe utilizar: **global variable**


Si sólo se va a utilizar la variable global (sin modificarla, no es necesario indicar lo anterior)

```python
#Uso de variable global dentro de una función
def f():
    #se puede utilizar el valor de la variable global g, si no se modifica
    a = g * 5
    #g = 5 #Si se incluye esta línea, daría un error porque se entiende g
           #local y en la línea anterior se usa sin tener un valor asignado
    print(g)
    print(a)

g = 3
f()
#print(a)
print(g)
```

```python
#Si se asigna y usa el mismo nombre de una variable global para una variable
#local, sólo existirá la local dentro de la función
def f():
    g = 4      # se crea una variable local para la función f
    g = g * 5  #se toma el valor de la variable local para modificarla
    print(g)

g = 3
f()
print(g)
```

## 6. Ámbitos anidados (*nested scopes*)

Los ámbitos anidados aparecen cuando dentro de una función se definen otras funciones. Conceptos de interés:
* **Ámbito global:** define todas las variables y funciones definidas en el espacio general de nombres del intérprete de Python, excluyendo las variable y funciones de funciones
* **Ámbito local:** define las variables y funciones definidas dentro de una función particular, que sólo tendrán significado dentro de ese ámbito y no se visualizan fuera de él
* **Conexión entre ámbitos:** *Python busca una variable o función en el ámbito local en el que se encuentre ejecutándose y si no la encuentra, intentará localizarla en el ámbito inmediatamente superior*


* **En una función se puede definir otra función, válida sólo en el ámbito en el que se define:**

```python
def f(x):
    def g():
        x = 'abc'
        print('x =',x)
    g()
    print('x =', x)
    return x
f(4)
```

```python
g()  #función creada dentro del ámbito de la definición de f(x).
#Sólo se puede ejecutar desde el ámbito en el que es visible
```

## 10. Funciones *lambda*

Son funciones sin un nombre asociado y definidas con una notación simplificada, con una única expresión:
* Forma general: **lambda entrada: salida**
* Son **funciones anónimas (no es necesario darles un nombre) de una sola línea**, que se podrán usar en cualquier lugar en que una función sea necesaria. Funciones como sorted, map, filter tienen una función como argumento.
* Muy usadas para **pasar funciones como parámetros**.
* Permiten reducir código, admiten varios parámetros de entrada y permiten devolver varios valores de salida sin utilizar *return*
* Forma general: **lambda parámetros_entrada: valores_a_devolver**

```python
help(sorted)
```

```python
L = [(2,1), (4,3), (1,3), (6,2), (0,5)]
sorted(L) #ordena por la primera componente del iterable, por defecto
```

```python
L = [(2,1,3), (4,3,2), (1,3,0),(0,3,0), (6,2,9), (0,5,4)]
sorted(L, key=lambda x: x[-1])#indica que la ordenación se realice por la última
                              #componente del iterable, no analiza el resto de componentes
```
