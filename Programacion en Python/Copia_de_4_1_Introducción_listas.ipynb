## 1. Definición

Las listas son **estructuras de datos** formadas por una secuencia de elementos (valores numéricos, cadenas de texto, ..., del mismo o diferente tipo).

Para definir una lista se utilizarán los corchetes. Los elementos de una lista se separan por comas.

Las listas son los tipos de datos **más flexibles** de Python.

**Ejemplos de definición de listas:**

L0 = [ ] es una lista vacía

L1 = [0, 1, 2, 3]

L2 = ['A', 'B', 'C', 'D']

L3 = ['1', 'A', 2, 3]

**Operaciones básicas con listas: Suma, producto y comparaciones de listas:**

````python
#Suma de listas
L1 = ['a', 'b', 'c']
L2 = ['A', 'B']
L3 = L1 + L2
print(L3)
L3 = L2 + L1
print(L3)
````

````python
#Suma de listas
print(L1 + ['*'] + L2)
````

````python
#Producto sobre una lista:
L1 = ['a', 'b', 'c']
print(L1*3)
L1 = [1,3,5]
print(L1*3)
````

````python
#Comparaciones de listas:
L1 = ['a', 'b', 'c']
L2 = ['a',  'b', 'c']
if L1 == L2:
    print("Listas con los mismos valores")
print("Fuera")
````

````python
#Comparaciones de listas:
L1 = ['a', 'b', 'c']
L2 = ['a', 'b', 'c', 1]
if L1 == L2:
    print("Listas con los mismos valores")
else:
    print("Listas con distintos valores")
````

````python
#Comparaciones de listas elemento a elemento:
L1 = [0, 1, 2, 3]
L2 = [5, 6, 9]
if L1 < L2:
    print("L1 menor")
````

````python
#Comparaciones de listas elemento a elemento:
L1 = [0, 1, 2, 3]
L2 = [0, 1, 1]
if L1 < L2:
    print("L1 menor")
````

### B. Indexado para acceder al contenido

Permite acceder a una o varias posiciones de la lista

Acceder a elementos de una lista por índice:

```mi_lista[index]```

Acceder al último elemento de una lista:

```mi_lista[-1]```

Acceder a una sublista de la lista original:

```mi_lista[i:j]```

Acceder a una sublista de la lista original especificando un paso entre elementos:

```mi_lista[i:j:step]```

Lista en sentido inverso:

```mi_lista[::-1]```
También se puede conseguir mediante:
```reversed(mi_lista)```



````python
lista = [3, 6, 'a', 3.5]
print(lista[0])
print(lista[-1])
print(lista[1:3])
print(lista[0:-1:2])
print(lista[::2])
print(lista[::])
print(lista[::-1])
print(lista[-2])
````

````python
mi_lista = [1, 2, 3, 4, 5]
mi_lista = [10, 23, 31, 14, 15]

print(mi_lista[0])

print(mi_lista[-1])

print(mi_lista[2:4])
print(mi_lista[0:4:2])
print(mi_lista[4:0:-2])
print(mi_lista[::-1])
````

````python
mi_lista = list(range(0, 101, 2))
print(mi_lista)
print(mi_lista[-1])
````

````python
#Acceso fuera de rango

lista = [3, 6, 'a', 3.5]
print(lista[4])
````

### C. Mutabilidad. Métodos aplicados a listas

Los valores de la lista se pueden modificar a través de la asignación o también utilizando métodos

````python
lista = [3, 6, 'a', 3.5]
print(lista)
lista[0] = 5   #El operador de asignación sobre una posición de la lista para modificar su contenido
print(lista)

lista[1] = [4,6]
print(lista)

lista[2] = "cadena"
print(lista)
print(lista[2][1])
````

````python
lista = [3, 6, 'a', 3.5]
print(lista)
lista[1] = [4,6]
print(lista)
print(lista[1])
print(lista[1][1])
````

**Métodos que permiten modificar una lista:**

* ```lista.append(elemento)``` Agregar nuevo elemento al final de la lista
* ```lista.insert(index, elemento)``` Insertar elemento en un índice o posición
* ```lista.extend(lista2)``` Añade al final de mi_lista los elementos de lista2


* ```lista.remove(elemento)``` Borrar la primera aparición del elemento indicado como argumento de mi_lista
* ```lista.pop(index)``` Extraer el elemento que se encuentra en un índice o posición dada como argumento. EL ELEMENTO SE ELIMINA DE LA LISTA. Si no se indica una posición, se extrae el último elemento de la lista. Devuelve el valor extraído

````python
mi_lista = [1, 2, 3, 4, 5]

mi_lista.append(8)
print(mi_lista)
mi_lista.append([1,2])
print(mi_lista)
````

````python
mi_lista.insert(2, "abc")
print(mi_lista)
mi_lista.insert(24, "abc") # No da error indicar una posición inexistente
print(mi_lista)
mi_lista.insert(-1, "abc")
print(mi_lista)
````

````python
mi_lista.extend([1,1,1])
print(mi_lista)
````

````python
x = mi_lista.remove("abc") #El método devuelve None
print(mi_lista)
print(x)

x = mi_lista.pop(3)        #El método devuelve el elemento extraído
print(x)
print(mi_lista)
````

**Otros métodos para operar con listas**

* ```lista.index(elemento)``` Devuelve el primer índice (posición) en la que se encuentra un elemento de la lista dado como argumento. La posición se cuenta desde la 0.
* ```lista.count(elemento)``` Devuelve el número de veces que se repite **elemento** en la lista.

* ```lista.copy()``` Devuelve una copia de la lista original.

````python
print(mi_lista.index(1))
print(mi_lista.index(1))
````

````python
print(mi_lista.count(1))
````

````python
lista = mi_lista.copy()
print(lista)
lista.pop()
print(lista)
print(mi_lista)
````

### F. Diferencias entre métodos y funciones
Algunos métodos y funciones realizan una operación similar, aunque existen diferencias en su aplicación y resultados

````python
help(sorted) #función de ordenación que devuelve una nueva lista ordenada
#sin modificar la dada como argumento
````

````python
#Aplicación de la función sorted()
mi_lista = [7, 9, 5, 7, 10, 7, 5, 60, 8, 5]
lista_ord = sorted(mi_lista)   #Ordena ascendentemente,
print(lista_ord)               #devolviendo la nueva lista
lista_ord_des = sorted(mi_lista,reverse=True)   #Ordena descendentemente
print(lista_ord_des)
print(mi_lista)
````

````python
help(mi_lista.sort) #método de ordenación que se aplica sobre una lista,
#modificándola y devolviendo None
````

````python
#Aplicación del método lista.sort()
mi_lista = [7, 9, 5, 7, 10, 7, 5, 60, 8, 5]
mi_lista.sort()  #método para ordenación
print(mi_lista)
````

````python
mi_lista.sort(reverse=True)

print(mi_lista)
````

**Resumen:**

* Objeto.sort() - Es un MÉTODO. Modifica la lista original y devuelve None
* sorted() - Es una FUNCIÓN. No modifica la lista original y devuelve una nueva lista ordenada

Lo mismo ocurre con:
* Objeto.reverse() - Es un MÉTODO. Modifica la lista original dándole la vuelta  y devuelve None
* reversed() - Es una FUNCIÓN. Devuelve un tipo especial de objeto (*reverse iterador*) con los valores de la lista original dándoles la vuelta

*Los métodos van asociados a objetos de la clase a la que pertenecen. Los métodos se invocan sobre objetos:*

En la siguiente instrucción, la clase será las listas de python, y el objeto de esa clase es la lista mi_lista (una instancia particular de esa clase), y el método que invocamos sobre el objeto es .sort()

```mi_lista.sort()```

Las funciones no están asociadas a ningún objeto. Pueden invocarse simplemente por su nombre y aplicarse al argumento dado:

```sorted(mi_lista)```

````python
help(reversed)
````

````python
lista = [5,7,3]
lista_inversa = reversed(lista)   #la función devuelve un 'reverse iterator'
print(lista_inversa)   #es un 'iterador reverse'
print(*lista_inversa)  #permite ver los valores del iterador
print(list(reversed(lista)))    #convierte a una lista los elementos del iterador
````

````python
for i in reversed(lista):  #Uso del reverse iterator'
    print(i)
````

````python
for i in lista_inversa:  #Uso del reverse iterator'
    print(i)
````

````python
#Aplicación del método
lista = [5,7,3]
lista.reverse()
print(lista)
````

````python
#Para crear una nueva lista, o usarlo como iterador, se RECOMIENDA invertir la
#lista con el indexado
lista = [5,7,3]
lista_inversa = lista[::-1]
print(lista_inversa)
for i in lista[::-1]:
    print(i)
````

### G. Funciones y métodos de listas y *string* (```list, str, split, join```)

Es interesante conocer funciones y métodos que permiter trabajar con listas y cadenas de caracteres o *strings*

La función list() con un argumento de tipo *string* convierte a este en una lista con cada uno de los caracteres que componen el *string*

````python
list("abc")
````

````python
#Conversión de cadena en lista de caracteres de forma directa: list()
cadena = "Esto es una prueba"
lista = list(cadena)
print(lista)
````

La función str() con un argumento de tipo lista convierte a esta en una cadena de caracteres. Convierte cualquier otro elemento a *string*

````python
str(['a', 'b', 'c'])
````

````python
str(123)
````

El método *split()* se aplica sobre un *string* separando ciertos elementos mediante el separador indicado como argumento. Devuelve una lista con todo los elementos separados. *maxsplit* indica el número de veces que se aplicará el separador, en caso de no indicarlo, se aplicará a todo el string.

 ```string.split(separador,maxsplit)``` Separa los elementos del *string* de acuerdo al separador y devuelve una lista con todos ellos.

````python
# División de una cadena en lista considerando algún separador: cadena.split()
cadena = "Esto es una prueba"
lista = cadena.split()
#el método split considera como elemento separador el espacio en blanco
print(lista)
````

````python
#cadena.split(separador)
"Esto es una prueba".split(" ")
````

````python
cadena = "Esto es una prueba, pero determinante"
lista = cadena.split(',')
print(lista)
````

````python
lista[1]
````

````python
cadena = "Pilar, Lucas, Carla, Irene"
lista = cadena.split(", ")
print(lista)
lista = cadena.split(", ",2)
print(lista)
````

````python
help(cadena.split)
````

El método *join* se aplica sobre una lista para unir todos sus elementos para conformar un *string* utilizando un separador dado entre dichos elementos. También se puede aplicar sobre cualquier iterable dado como argumento:

 ```separador.join(iterable)``` Une los elementos del iterable y devuelve una cadena de texto utilizando el separador indicado entre cada elemento.

````python
#delimitador.join(iterable)
lista1=['a','b','c']
print("-".join(lista1))
x = "".join(lista1)    #asociando el *string* resultante a una variable
print(x)
type(x)
````

````python
#Unión de los elementos de una lista de strings en un único string: delimitador.join()
lista = ['Esto', 'es', 'una', 'prueba']
separador = '-'
cadena = separador.join(lista)
print(cadena)
type(cadena)
cadena = '*'.join(lista)
print(cadena)
````

````python
lista = [4, 7, 7, 3]
cadena = '*'.join(lista)  #Los elementos de la lista deben ser str para el join
````

````python
#Unión de los elementos de tipo numérico de una lista en un string: antes debe
#convertirse cada número en string
lista = [4, 7, 7, 3]
L = []
for i in lista:
    L.append(str(i))
print(L)
cadena = '*'.join(L)
print(cadena)
````

````python
help("".join)
````
